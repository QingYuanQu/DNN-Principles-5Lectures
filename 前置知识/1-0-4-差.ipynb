### 1.2 差

|                  | 数据本身（基于数据分布/均值） | 理论（基于真实值）| 模型（基于预测值）|
| ---------------- | ---------------------------- | ---------------------------- | ---------------------------- |
| 差               | 离差（$x_i-\bar{x}$）| 误差（$x_i-x_{true}$）| 残差（$x_i-\hat{x}_i$）|
| 绝对差           | 绝对离差（$|x_i-\bar{x}|$）| 绝对误差（$|x_i-x_{true}|$）| 绝对残差（$|x_i-\hat{x}_i|$）|
| 平方差           | 离差平方（$(x_i-\bar{x})^2$） | 误差平方（$(x_i-x_{true})^2$） | 残差平方（$(x_i-\hat{x}_i)^2$） |
| 平方差的和       | 离差平方和（$SST=\sum(x_i-\bar{x})^2$） | 误差平方和（$\sum(x_i-x_{true})^2$） | 残差平方和（$SSE=\sum(x_i-\hat{x}_i)^2$） |
| 平均平方差       | 方差（总体：$\sigma^2=\frac1n\sum(x_i-\bar{x})^2$；样本：$s^2=\frac1{n-1}\sum(x_i-\bar{x})^2$） | 均方误差（$MSE=\frac1n\sum(x_i-x_{true})^2$） | 均方残差（$MSE_{res}=\frac1{n-p}\sum(x_i-\hat{x}_i)^2$，$p$为模型参数个数） |
| 均方根平均平方差 | 标准差（$\sigma=\sqrt{\sigma^2}$；$s=\sqrt{s^2}$） | 均方根误差（$RMSE=\sqrt{MSE}$） | 均方根残差（$RMSE_{res}=\sqrt{MSE_{res}}$） |

- 离差的本质是**数据点与数据集均值的偏离**，属于对数据本身分布特征的描述，和“理论/模型”无关。
- 误差的本质是**观测值与理论真实值的偏离**，真实值通常不可知（比如物理实验的“真值”），是理论层面的概念。
- 残差的本质是**观测值与模型预测值的偏离**，是模型拟合效果的衡量指标，是应用层面的概念。


1.  **易混淆细节**
    - 方差的**自由度**：总体方差除以样本量$n$，样本方差除以$n-1$，目的是无偏估计；
    - 均方残差的自由度：除以$n-p$（$p$是模型参数个数），因为模型拟合会消耗自由度；
    - 残差平方和（SSE）是模型**未解释的变异**，离差平方和（SST）是数据**总变异**，二者的差值是模型**解释的变异（SSR）**，满足 $SST=SSR+SSE$。
